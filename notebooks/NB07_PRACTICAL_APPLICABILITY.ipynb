Public execution notebook for the shell-egg NIR reproducibility workflow. Scientific code is stored under `notebooks/src/`. Stored cell outputs, execution counts, and user-specific Colab runtime metadata are intentionally excluded from this wrapper.


```python
from pathlib import Path
import subprocess, sys

REPO_DIR = Path('/content/nir-eggs-repro')
if not (REPO_DIR / '.git').exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/Julians30/Using-a-Low-Cost-Portable-Near-Infrared-NIR-and-Deep-Learning-Approaches-Based-on-RNN-and-LSTM-', str(REPO_DIR)], check=True)
else:
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull', '--ff-only'], check=True)

sys.path.insert(0, str(REPO_DIR))
```

```python
SOURCE_FILES = [
    REPO_DIR / 'notebooks/src/NB07_part01.py',
    REPO_DIR / 'notebooks/src/NB07_part02.py',
    REPO_DIR / 'notebooks/src/NB07_part03.py',
]

for source_file in SOURCE_FILES:
    print(f'Executing: {source_file.relative_to(REPO_DIR)}')
    source = source_file.read_text(encoding='utf-8')
    exec(compile(source, str(source_file), 'exec'), globals())
```
